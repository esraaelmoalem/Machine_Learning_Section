```python

import numpy as np
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsClassifier

students = np.array([
    [1, 2],
    [2, 1],
    [2, 3],
    [3, 2],
    [6, 5],
    [7, 7],
    [8, 8],
    [7, 5]
])

results = ['Fail', 'Fail', 'Fail', 'Fail',
           'Pass', 'Pass', 'Pass', 'Pass']

K = 3

model = KNeighborsClassifier(n_neighbors=K)
model.fit(students, results)

new_student = np.array([[5, 4]])

prediction = model.predict(new_student)

distances, indexes = model.kneighbors(new_student)

print("Prediction:", prediction[0])

print("\nNearest Neighbors:")

for i in indexes[0]:
    print(students[i], results[i])

print("\nDistances:")
for d in distances[0]:
    print(round(d, 2))

# Drawing
plt.figure(figsize=(9, 6))

for i in range(len(students)):
    if results[i] == 'Pass':
        plt.scatter(students[i][0], students[i][1],
                    color='green', label='Pass' if i == 4 else "")
    else:
        plt.scatter(students[i][0], students[i][1],
                    color='red', label='Fail' if i == 0 else "")

    plt.text(students[i][0] + 0.1,
             students[i][1] + 0.1, f'P{i+1}')

plt.scatter(5, 4, color='blue', marker='*',
            s=250, label='New Student')

for i in indexes[0]:
    plt.plot(
        [5, students[i][0]],
        [4, students[i][1]],
        'b--'
    )

plt.title("KNN - Student Pass or Fail Prediction")
plt.xlabel("Study Hours")
plt.ylabel("Attendance")
plt.legend()
plt.grid()
plt.show()
```
